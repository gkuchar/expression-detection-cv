# Preprocess Data:
## 1. Grayscale
## 2. Square and Center Around Face
## 3. Normalize Pixel Values

**Input:** `data/raw_test`, `data/raw_train`

**Output:** `data/test`, `data/train`

## Load Files and Imports

```python
# imports
import cv2
import numpy as np
from pathlib import Path

# file paths
input_train_path = '../data/raw_train'
input_test_path = '../data/raw_test'

output_train_path = '../data/train'
output_test_path = '../data/test'

inputs = [input_train_path, input_test_path]
outputs = [output_train_path, output_test_path]
```

```python
# Pixel Normalization Function
def normalize(img: np.ndarray) -> np.ndarray:
    min_val = img.min()
    max_val = img.max()
    denom = max_val - min_val if max_val != min_val else 1.0
    normalized = (img - min_val) / denom        # [0, 1] float
    return (normalized * 255).astype(np.uint8)  # back to uint8 for saving
```

## Preprocess: Grey + Square + Normalize

```python
size = 128 # pixels

# for both train and test input folders
for i in range(len(inputs)):
    input_path = inputs[i]
    output_path = outputs[i]
    for emotion_dir in Path(input_path).iterdir(): # for each emotion subdir
        for img_path in emotion_dir.rglob("*.jpg"): # preprocess each image
            img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
            img = cv2.resize(img, (size, size))
            img = normalize(img)

            out_path = Path(output_path) / img_path.relative_to(input_path)
            out_path.parent.mkdir(parents=True, exist_ok=True)
            cv2.imwrite(str(out_path), img)
```
